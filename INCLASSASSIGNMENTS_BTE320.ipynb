design simple numerical calculator in python. user provides 3 inputs: two numbers and mathematical operator. program should print output

```python
N1 = (float(input("Numer 1: ")))
N2 = (float(input("Numer 2: ")))
M = ((input("Mathematical Operator: ")))
res = "None"

if M == "+":
  res = N1 + N2
elif M == "-":
  res = N1 - N2
elif M == "*":
  res = N1 * N2
elif M == "/":
  res = N1 / N2
elif M == "//":
  res = N1 // N2
elif M == "**":
  res = N1 ** N2
elif M == "%":
  res = N1 % N2
else:
  print("Invalid Operation")
print(f"Result:", res)
```

```python
import math

print("Ad\tProfit")
tp = 10
standard = 20

for ad in range(0, 225, 25):
  extra = 2 * round(math.sqrt(ad))
  revenue = (extra * tp) + (standard * tp)
  profit = revenue - (200 + ad)
  print(f"${float(ad)}\t${float(profit)}")
```

create recursive function for a string to be reversed

```python
def rev(s):
  if len(s) < 2:
    return s
  else:
    return s[-1] + rev(s[:-1])

print(rev("1234567"))
```

```python
def roi(cost, revenue):
  assert cost != 0, "Cost must not equal 0!"
  assert isinstance(cost, (int, float)), "Cost must be a number!"
  assert isinstance(revenue, (int, float)), "Revenue must be a number!"

  return float((revenue-cost) / cost)

cost = float(input("Enter cost value: "))
revenue = float(input("Enter revenue value: "))

print(roi(cost, revenue))
```

```python
def loan_payment(principal, annual_rate, months):
  assert isinstance (principal, (int, float))
  assert isinstance (annual_rate, (int, float))
  assert isinstance (months, (int, float))


  P = principal
  n = months
  r = annual_rate / 12

  try:
    if r !=0:
      pmt = r * P / (1-(1+r) ** (-n))
    else:
      pmt = P / n
  except:
    raise ValueError("Functions inputs must be positive!")

  return pmt

principal = float(input("Input principal here: "))
annual_rate = float(input("Input annual rate here: "))
months = float(input("Input term months here: "))

print(loan_payment(principal, annual_rate, months))
```

```python
class Greeting:
  def __init__(self, first, last):
    self.first = first
    self.last = last

  def display(self, formal=False):
    if formal:
      print(f"{self.last}, {self.first}")
    else:
      print(self.first)

  def __str__(self):
    return "Instance of class 'Gretting'"

name = Greeting("Alexander", "Carlson")
print(name)
name.display()
name.display(formal=True)
```

```python
class Student:
  def __init__(self, name, number):
    self.__name = name
    self.number = number
    self.__courses = []

  def get_name(self):
    return self.__name

  def get_courses(self):
    return self.__courses

  def set_name(self, newName):
    self.__name = newName

  def __len__(self):
    return len(self.__courses)

  def enroll(self, newCourse):
    if newCourse not in self.__courses:
      self.__courses.append(newCourse)
      print(f"Student {self.__name} succesfully enrolled in {newCourse}")
    else:
      print(f"Student {self.__name} is already enrolled in {newCourse}")


s = Student('Alex Carlson', 12345)
print(f'Student name {s.get_name()}')
print(f'Show currently enrolled courses  {s.get_courses()}')
s.enroll("BTE 320")
print(f'Show currently enrolled courses  {s.get_courses()}')
s.enroll("BTE 320")
print(f'Show currently enrolled courses  {s.get_courses()}')
```

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y


  def distance(self, other): #when you have method in classes that take input beyond self automatically make other
    dx = (self.x - other.x) ** 2
    dy = (self.y - other.y) ** 2
    d = (dx + dy) ** 0.5

class Rectangle:
  def __init__(self, p1, p2, p3, p4):
    self.p1 = p1
    self.p2 = p2
    self.p3 = p3
    self.p4 = p4
    self.width = self.width_()
    self.height = self.height_()

  def width_(self):
    return self.p1.distance(self.p2)

  def height_(self):
    return self.p1.distance(self.p3)

  def area(self):
    return self.width * self.height

p1 = Point(0, 0)
p2 = Point(5, 0)
p3 = Point(0, 5)
p4 = Point(5, 5)

z = Rectangle(p1, p2, p3, p4)
z.area()
```

```python
class Employee:
  def __init__(self, name, wage, hours):
    self.name = name
    self.__hourly_wage = wage
    self.hours = hours

  def get_wage(self):
    return self.__hourly_wage

  def set_wage(self, newWage):
    if newWage <= 0:
      print("Invalid wage value.")
    else:
      self.__hourly_wage = newWage

  def calculate_pay(self):
    return self.__hourly_wage * self.hours

  def __str__(self):
    return f"Employee: {self.name}\nHourly wage: {self.__hourly_wage}"

class Manager(Employee):
  def __init__(self, name, wage, hours, bonus):
    super().__init__(name, wage, hours)
    self.bonus = bonus # decimal format, e.g., 0.2 for 20%

  def calculate_pay(self):
    return (1 + self.bonus) * self.get_wage() * self.hours


employee = Employee("Bob", 50, 10)
manager = Manager('Alice', 100, 15, 0.2)
print(employee)
print(manager)
print()
print(f"Employee {employee.name} weekly payment is {employee.calculate_pay()}$")
print(f"Manager {manager.name} weekly payment is {manager.calculate_pay()}$ that includes a {manager.bonus * 100}% bonus.")
```
